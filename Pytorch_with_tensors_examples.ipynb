**Basic Tensor Operations - Creating, Reshaping, and Inspecting Tensors**

```python
import torch

#Create Tensors
a = torch.tensor([1,2,3]) #1D Tensor
b = torch.tensor([[1, 2], [3, 4]]) #2D Tensor

#Inspect Tensors
print(a.shape)
print(b.shape)
print("Data Type", b.dtype)
print("Device", b.device)

#Reshaping
c = torch.arange(1, 10).reshape(3, 3)
print("Reshaped Tensor:\n", c)
```

**Elementwise Operations - Addition, Multiplication, and Broadcasting**

```python
x = torch.tensor([1, 2, 3])
y = torch.tensor([4, 5, 6])

print("Elementwise addition: ", x + y )
print("Elementwise multiplication: ", x * y )

# Broadcasting (adding scalar to tensor)
z = torch.tensor([[1,2], [3, 4]])
print("Broadcast Addition:\n", z + 5)
```

**Matrix Operations - Dot Product, Matrix Multiplication, Transpose, Inverse**

```python
# Matrix multiplication

A = torch.tensor([[1, 2], [3, 4]])
B = torch.tensor([[5, 6], [7, 8]])

dot_product = torch.dot(torch.tensor([1, 2]), torch.tensor([3, 4])) # Dot product (1D tensors)
print("Dot Product:", dot_product)
matmul = torch.matmul(A, B) # Matrix multiplication
print("Matrix Multiplication:\n", matmul)


```
